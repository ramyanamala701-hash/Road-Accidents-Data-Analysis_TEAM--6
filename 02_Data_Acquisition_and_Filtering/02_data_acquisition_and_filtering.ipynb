# Stage 2 — Data Acquisition & Filtering

## Objective

Demonstrate optional filters using values present in the source. The default pipeline retains all rows.

## Source and scope

Input: `data/raw/india_road_accidents_dataset.csv` (the provided 240 × 25 accident-level CSV).

## Interpretation note

Examples are subsets of this CSV, not representative samples.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display
ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "src").is_dir()), None)
if ROOT is None:
    raise RuntimeError("Run this notebook from the project root or a child folder.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from src.data_loader import load_dataset, find_dataset, inspect_dataset
```

```python
from src.data_acquisition import filter_records
raw_df = load_dataset()
state_example = raw_df.state_ut.iloc[0]
severity_example = raw_df.severity.iloc[0]
road_example = raw_df.road_type.iloc[0]
weather_example = raw_df.weather.iloc[0]
examples = {
    "latest observed year": filter_records(raw_df, years=[int(raw_df.year.max())]),
    f"State/UT: {state_example}": filter_records(raw_df, locations=[state_example]),
    f"severity: {severity_example}": filter_records(raw_df, severities=[severity_example]),
    "fatal accidents": filter_records(raw_df, fatal_only=True),
    "pedestrians involved": filter_records(raw_df, pedestrians=True),
    "alcohol involved": filter_records(raw_df, alcohol=True),
    f"road type: {road_example}": filter_records(raw_df, road_types=[road_example]),
    f"weather: {weather_example}": filter_records(raw_df, weather=[weather_example]),
}
for label, subset in examples.items():
    print(f"{label}: {len(subset)} of {len(raw_df)} records")
    display(subset.head(3))
```
